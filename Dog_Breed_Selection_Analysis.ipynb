## Problem Statement

The objective of this project was to identify suitable dog breeds for an individual seeking a male puppy that fits specific lifestyle preferences.

The analysis was based on a dataset containing information on 97 dog breeds. The selection criteria were:

- **Height:** Below-average maximum male height.
- **Barking tendency:** A barking score of 1 or 2.
- **Compatibility with children:** A good-with-children score of 4 or 5.

The analysis used Python and Pandas to filter the dataset and identify breeds that met all three criteria.

```python
average_height = dog_breeds['max_height_male'].mean()
print ("Average maximum male height:", average_height)
```

```python
suitable_dogs = dog_breeds[(dog_breeds['max_height_male'] < average_height) & (dog_breeds['barking'].isin([1, 2])) & (dog_breeds['good_with_children'].isin([4, 5]))]
suitable_dogs
```

```python
columns_to_keep = ['max_height_male', 'barking', 'good_with_children']
print ("Number of sitable breeds:", len(suitable_dogs))
suitable_dogs[columns_to_keep]
```
