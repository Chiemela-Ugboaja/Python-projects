USING FUNCTION TO MINIMIZE CODES

```python
def play_one_game():
    coin = np.random.choice(['Heads', 'Tails'])
    dice = np.random.randint(1, 7)

    if coin == 'Heads' and dice <= 3:
        return 'Ruth'
    elif coin == 'Heads' and dice > 3:
        return 'Nkechi'
    elif coin == 'Tails' and dice == 1:
        return 'Ogechi'
    elif coin == 'Tails' and dice >= 2:
        return 'Collins'

def simulate(n=1000):
    results = [play_one_game() for _ in range(n)]
    return pd.Series(results).value_counts()

print(simulate(1000))
```

```python
for name, count in pd.Series(results).value_counts().items():
    print(name, count)
```
